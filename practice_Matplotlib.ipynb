```python
import pandas as pd
from matplotlib import pyplot as plt
from matplotlib import style
```

```python
dt = pd.read_csv('spf.csv')
```

```python
df = dt.sample(3600)
```

```python
display(df)
```

```python
df.isna().sum()
```

```python

df.drop('Parental_Education_Level', axis=1, inplace=True)
df.drop('Distance_from_Home', axis=1, inplace=True)


display(df)
```

```python
df.reset_index(inplace=True)
```

```python
df
```

```python
display(df)
```

```python
df.drop('level_0', axis=1, inplace=True)
```

```python
df.drop('index', axis=1, inplace=True)
```

```python
df
```

## Question to answer

```python
## Does the internet affect the students study hour 

df.Internet_Access.value_counts()
```

```python
# Do peer influence affect exam score


peer = df.groupby('Peer_Influence')['Exam_Score'].mean()

peer
```

```python
plt.figure(figsize=(3, 4))
style.use('ggplot')
plt.bar(
    peer.index,
    peer.values,
    color=['red', 'blue', 'green'],
    edgecolor= 'black',
    width=0.95
)

plt.title('Average exam score by peer influence')
plt.xlabel('Peer influence')
plt.ylabel('Exam score')
plt.show()


# just some slight effect
```

```python
# Do school type affect exam score 

st = df.groupby('School_Type')['Exam_Score'].mean()
```

```python
plt.figure(figsize=(2, 4))
style.use('ggplot')
plt.bar(
    st.index,
    st.values,
    color=['orange', 'violet'],
    width=0.95,
    edgecolor= 'black'
)

plt.title('Average Exam score by School type')
plt.xlabel('School type')
plt.ylabel('Exam score')
plt.show()

# No significant effect 
```

```python
# Does family income affect exam score?
ft = df.groupby('Family_Income')['Exam_Score'].mean()
ft
```

```python
plt.figure(figsize=(3,4))
plt.bar(
    ft.index,
    ft.values,
    color=['orange', 'violet', 'yellow'],
    edgecolor= 'red',
    width=0.95
)

plt.title('Average Exam score by Family income')
plt.xlabel('Family income')
plt.ylabel('Exam score')
plt.show()


# there is no significant effect from high and medium family income on exam score, but slight effect on low family income
```

```python
# Dose access to internet affect exam score

net = df.groupby('Internet_Access')['Exam_Score'].mean()
net
```

```python
plt.figure(figsize=(3,4))
plt.bar(
    net.index,
    net.values,
    color=['gold', 'indigo'],
    edgecolor='blue',
    width=0.95
)

plt.title('Average exam score to student access to the internet')
plt.xlabel('Access to internet')
plt.ylabel('Exam score')
plt.show()


# We see that access to the internet slight improve the students score, then removal of internet
```

```python
plt.figure(figsize=(8,8))
plt.scatter(df['Previous_Scores'], df['Exam_Score'], alpha=0.4, color='blue')
plt.title('Pre-Exam scores VS Exam scores')
plt.xlabel('Pre-Exam scores')
plt.ylabel('Exam scores')
plt.show()



# checking for correletion between previous scores and recent scores
```
