```python
import pandas as pd
from matplotlib import pyplot as plt
from matplotlib import style
import seaborn as sns
```

```python
sns.set_theme(style='whitegrid')
```

```python
df= pd.read_csv('spf.csv')
```

```python
df.info()
```

```python
sns.scatterplot(data=df, x='Hours_Studied', y='Exam_Score', hue='Gender')

plt.title('hours studied vs exam score')
plt.show()
```

```python
sns.set_theme(style='darkgrid')
sns.scatterplot(data=df, x='Attendance', y='Exam_Score', hue='Motivation_Level')

plt.title('attendance vs exam score')
plt.show()
```

```python
sns.barplot(data=df, x='School_Type', y='Exam_Score', hue='Exam_Score')
plt.title('Exam score vs school type')
plt.show()
sns.set_theme(style='ticks')
```

```python
sns.barplot(data=df, x='Parental_Involvement', y='Exam_Score', hue='Gender')
plt.title('Exam score vs Parent involvement')
plt.show()
sns.set_theme(style='ticks')
```

```python
sns.histplot(data=df, x='Exam_Score', bins=50, kde=True)
plt.title('Exam score distribution')
plt.show()
```

```python
sns.histplot(data=df, x='Hours_Studied', hue='Gender', bins=30, kde=True)
plt.title('Studied hours distribution')
plt.show()
```

```python
sns.boxplot(data=df, x='Motivation_Level', y='Exam_Score')
plt.title('Boxplot for Exam score Vs motivational level')
plt.show()
```

```python
sns.boxplot(data=df, x='School_Type', y='Exam_Score')
plt.title('Boxplot for Exam score Vs School Type')
plt.show()
```

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Heat map for numeric values')
plt.show()
```

```python
sns.set_theme(style='ticks')
sns.pairplot(df[['Exam_Score', 'Sleep_Hours', 'Attendance', 'Hours_Studied']])
plt.title('Pair plot')
plt.show()
```
