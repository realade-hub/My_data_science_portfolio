```python
#Statistics measures

mean_post_GPA = df['Post_Semester_GPA'].mean()
mean_weekly_gen_hour = df['Weekly_GenAI_Hours'].mean()

std_post_gpa = df['Post_Semester_GPA'].std()
std_weekly_gen_hour = df['Weekly_GenAI_Hours'].std()

q1_post =  df['Post_Semester_GPA'].quantile(0.25)
q3_post =  df['Post_Semester_GPA'].quantile(0.75)
IQR_post = q1_post - q3_post

q1_week = df['Weekly_GenAI_Hours'].quantile(0.25)
q3_week = df['Weekly_GenAI_Hours'].quantile(0.75)
IQR_week = q1_week - q3_week

print(f'The mean of post semster GPA is : {mean_post_GPA: .3f}')
print(f'The standard deviation for Post semster GPA is : {std_post_gpa: .3f}')
print(f'The mean of weekly GenAi use per hour is : {mean_weekly_gen_hour: .3f}')
print(f'The standard deviation for weekly GenAi use per hour is : {std_weekly_gen_hour: .3f}')
print(f'The IQR for post semster GPA is {IQR_post: .3f}')
print(f'The IQR for weekly GenAi use is {IQR_week: .3f}')
```

```python
from scipy.stats import ttest_1samp, ttest_ind, chi2_contingency, pearsonr, spearmanr
```

```python
#Chi square 
con_table = pd.crosstab(df['Institutional_Policy'], df['Paid_Subscription'])
chi2, p_value, dof, expect = chi2_contingency(con_table)
print(f'chi2: {chi2: .3f}')
print(f'P-value: {p_value: .3f}')
```

```python
#correlatiom

r, p_value = pearsonr(df['Weekly_GenAI_Hours'], df['Post_Semester_GPA'])

print(f'Pearson r: {r: .3f}, P-value: {p_value: .3f}')
print(f'the Pearson r: {r: .3f} is negetive meaning that there is an inverse or negetive reletionship')
```

```python
r, p_value = pearsonr(df['Pre_Semester_GPA'], df['Post_Semester_GPA'])
print(f'Pearson r: {r: .3f}, P-value: {p_value: .3f}')
print(f'the Pearson r: {r: .3f} is postive and closer to 1, meaning that there is very strong reletionship between them')
```

```python
rho, p_value = spearmanr(df['Traditional_Study_Hours'], df['Skill_Retention_Score'])
print(f'Spearmanr rho: {rho: .3f}, P-value: {p_value: .3f}')
print(f'Rho ={rho: .3f}, this mean that there is weak reletionship')
```
