```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('spf.csv')
```

```python
df
```

```python
x = df.Hours_Studied
y = df.Exam_Score
```

```python
from matplotlib import style 
```

```python
plt.figure(figsize=(16, 16))
plt.plot(x,y)
style.use('ggplot')
plt.show()

# we can call it a linear relationship, the higher the time studied the higher the score
```

```python
plt.scatter(x, y, alpha=0.5, color= 'blue')
plt.show()

# same as the line chart
```

```python
x = df.Exam_Score
plt.hist(x, bins=15, edgecolor='blue', color='pink')
plt.show()
```

```python
ml = df.groupby('Motivation_Level')['Exam_Score'].mean()
plt.figure(figsize=(8, 6))
plt.bar(
    ml.index,
    ml.values,
    color=['red', 'purple', 'violet'],
    edgecolor='black'
)
plt.title('Average exam score by motivation level')
plt.xlabel('Motivation Level')
plt.ylabel('Average exam score')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10,6))
ax1.scatter(df['Hours_Studied'], df['Exam_Score'], alpha=0.5, color='blue')
ax1.set_title('Hours studied VS Exam score')
ax1.set_xlabel('hours studied')
ax1.set_ylabel('exam scores')

ax2.hist(df['Exam_Score'], bins=15, edgecolor='black', color='pink')
ax2.set_title('Distribution of exam score')
ax2.set_xlabel('Exam score')
ax2.set_ylabel('Frequency')

plt.suptitle('Students performance analysis', fontsize=26)
plt.tight_layout()
plt.show()
```
